```python
params = {
  "q": "Lagos 'food prices'",
  "hl": "en",
  "gl": "ng",
  "tbm" : "nws",
  "start": "0",
  "num": "100",
  "google_domain": "google.com",
  "tbs": "cdr:1,cd_min:1/1/2000,cd_max:6/30/2024",
  "api_key": key
}

search = GoogleSearch(params)
results = search.get_dict()

result_news_list=[]
results_full_list=[]

while results["search_information"]["news_results_state"] != 'Fully empty':
    result_news_list.append(results["news_results"])
    results_full_list.append(results)
    params["start"] = str(int(params["start"]) + 100)
    search = GoogleSearch(params)
    results = search.get_dict() 
    print(f"{params['start']}: {results['search_information']['news_results_state']}")
```

```python
results = np.concatenate(result_news_list)
len(results)
```
