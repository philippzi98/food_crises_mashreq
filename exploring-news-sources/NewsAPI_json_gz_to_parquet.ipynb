```python
import gzip
import json
from tqdm import tqdm
import pandas as pd
from datetime import datetime
import pyarrow.parquet as pq
import pyarrow as pa
import os
```

```python
input_path = "../../data/final/downloads_newsapi/raw/"
output_path = "../../data/final/downloads_newsapi/clean_parquet/"
```

```python
file_list = os.listdir(input_path)
file_list = [file for file in file_list if file.endswith(".json.gz")]
```

```python
for file_name in file_list:
    for year in range(2014,2025):
        article_list = []
        print(f"Processing {file_name} for year {year}")
        
        with gzip.open(input_path + file_name, "rt", encoding="utf-8", errors="ignore") as f:
            for line in tqdm(f):
                data = json.loads(line)
                
                # Check if the article is from the year we want
                date = datetime.strptime(data["date"], '%Y-%m-%d')
                if date.year == year:
                    # If the article is from the year we want, we add it to the list
                    article_list.append(pd.json_normalize(data))
        
        # Create a parquet file with the articles
        table = pa.Table.from_pandas(pd.concat(article_list))
        pq.write_table(table, output_path + f'{file_name.split(".")[0]}_{year}.parquet')
```
