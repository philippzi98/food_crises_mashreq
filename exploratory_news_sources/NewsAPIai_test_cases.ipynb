### Yambio - South Sudan

1. 06/16/2024 to 06/30/2024 - South Sudan - "Yambio" - English articles only

```python
query = {
  "$query": {
    "$and": [
      {
        "keyword": "Yambio",
        "keywordLoc": "body"
      },
      {
        "conceptUri": "http://en.wikipedia.org/wiki/South_Sudan"
      },
      {
        "dateStart": "2024-06-16",
        "dateEnd": "2024-06-30",
        "lang": "eng"
      }
    ]
  },
  "$filter": {
    "isDuplicate": "skipDuplicates"
  }
}

q = QueryArticlesIter.initWithComplexQuery(query)
# change maxItems to get the number of results that you want

q.count(er)
```

```python
article_list = []
for article in q.execQuery(er, maxItems=100):
    article_list.append(article)
    print(article)
```
